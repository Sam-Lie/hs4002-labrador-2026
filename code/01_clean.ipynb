## 4. Data Recoding and Transformation

This section recodes missing-value codes and transforms selected variables to prepare them for statistical analysis.

### 4.1 Recode WVS Missing-Value Codes

The WVS missing-value codes are recoded as `NaN` across all selected variables to distinguish missing responses from valid response categories.

```python
# Recode WVS missing-value codes as NaN across all selected variables
df_clean = df_selected.replace([-1, -2, -3, -4, -5], np.nan)
```

### 4.2 Check Variables After Missing-Value Recoding

The frequency distributions of all selected variables are examined to confirm that the WVS missing-value codes have been successfully recoded as `NaN`.

```python
for col in df_clean.columns:
    print("\n====================")
    print(col)
    print(df_clean[col].value_counts(dropna=False).sort_index())
```

### 4.3 Assess Reverse Coding for Q27–Q32

Q27–Q32 are reviewed to determine whether reverse coding is required. Q28–Q31 are already coded such that higher scores indicate more egalitarian gender-role attitudes, so no reverse coding is necessary. Q27 measures the importance of making one's parents proud rather than gender-role attitudes and is therefore not included in the gender-role scale. Q32 concerns the perceived fulfilment of being a housewife compared with working for pay and has an ambiguous conceptual direction, so it is not reverse-coded at this stage.

### 4.4 Reverse-Code Negatively Worded Science and Technology Items (Q160 to Q162)

Q160–Q162 are negatively worded in relation to the intended science and technology attitude scale. The items are reverse-coded so that higher scores consistently represent more positive attitudes toward science and technology.

```python
# Reverse-code Q160–Q162 so that the coding direction matches the intended interpretation
for col in ["Q160", "Q161", "Q162"]:
    df_clean[col] = df_clean[col].map(
        lambda x: 11 - x if pd.notna(x) else np.nan
    )
```

### 4.5 Check Q160–Q162 After Reverse Coding

The frequency distributions of Q160–Q162 are examined after reverse coding to verify that the response values have been recoded correctly and that higher scores consistently represent more positive attitudes toward science and technology.

```python
# Display frequencies and percentages for Q160–Q162
for col in ["Q160", "Q161", "Q162"]:
    print("\n====================")
    print(col)

    counts = df_clean[col].value_counts(dropna=False).sort_index()
    percentages = (counts / 2012 * 100).round(2)

    output = pd.DataFrame({
        "Frequency": counts,
        "Percentage of Full Sample (%)": percentages
    })

    output.index = output.index.map(
        lambda x: "Missing" if pd.isna(x) else x
    )

    print(output)
```

### 4.6 Assess Parental Education Coding (Q277R and Q278R)

Q277R and Q278R measure the highest educational level attained by the respondent's mother and father, respectively. The variables are recoded into three education groups, where 1 represents lower education, 2 represents middle education, and 3 represents higher education. The variables are already coded in ascending order of educational attainment, so no reverse coding is required.
